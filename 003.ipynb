```python
import csv, string, random

def generate_codes(team):
  codes = []
  for _ in range(150):
    random_string = ''.join(random.choices(string.ascii_lowercase, k=5))
    codes.append(f'{team}_{random_string}')
  return codes
```

```python
codes = generate_codes('ROMA')

with open('output/roma.csv', 'w', newline='') as f:
  writer = csv.writer(f)
  for code in codes:
    writer.writerow([code])
```

```python
codes = generate_codes('PICKERING')

with open('output/pickering.csv', 'w', newline='') as f:
  writer = csv.writer(f)
  for code in codes:
    writer.writerow([code])
```
